```python
import csv
with open("student data.csv","r") as file:
    reader=csv.reader(file)
    with open("students_with_grades.csv", "w", newline="") as outfile:
        writer = csv.writer(outfile)

        header = next(reader)
        header.append("Grade")
        writer.writerow(header)
    
        for row in reader:
          chem = int(row[2])
          phy = int(row[3])
          math = int(row[4])  
          total = chem + phy + math
          percentage = (total / 300) * 100  
          if percentage >=80:
               grade="A"
          elif percentage >= 70:
                grade = "B"
          elif percentage >= 60:
                grade = "C"
          elif percentage >= 50:
                grade = "D"
          else:
                grade = "F" 
          row.append(grade)
          writer.writerow(row)
          
   
```
