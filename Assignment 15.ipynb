1. List all cities where the area is greater than 5000, along with their population.

```python
import pandas as pd
import sqlite3 as sql
conn = sql.connect('factbook.db')
def sql_query(q):
    return pd.read_sql_query(q,conn)
q= "select name,population from facts  where Area>5000"
sql_query(q)
```

3. List the cities with a population growth rate between 1% and 5%, sorted in desc
by population growth.

```python
q = """SELECT c.name AS City_name
FROM facts f
INNER JOIN cities c
    ON f.id = c.facts_id
WHERE f.population_growth >= 1
  AND f.population_growth <= 5
ORDER BY f.population_growth DESC;
        


"""
sql_query(q)
```

4. Find countries that start with &quot;A&quot; and do not have any cities listed in the cities
table. The query should return the country name and city name. Using left join

```python
q="""SELECT f.name as country_name, c.name as city
FROM facts f 
LEFT JOIN cities c
    ON f.id = c.facts_id
WHERE f.name LIKE 'A%'
  AND c.name IS NULL;"""
sql_query(q)
```
