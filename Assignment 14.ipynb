#### **1. What is the average TotalPayBenefits for each JobTitle? exclude ('Not Provided', 'Not provided')**

```python
import pandas as pd
import sqlite3 as sql
conn = sql.connect('database.sqlite')
pd.read_sql_query("""select jobtitle , AVG(totalpaybenefits) from salaries where jobtitle not in ('Not provided','Not Provided')
group by jobtitle""",conn)
```

#### **2. Show The JobTitle and Overtime Pay for All Employees With Overtime Pay greater than 50000**

```python
pd.read_sql_query("""select jobtitle,OvertimePay from salaries where OvertimePay  >50000  
""",conn)
```

#### **3. What is the total number of unique employees? exclude ('Not Provided', 'Not provided')**

```python
pd.read_sql_query("""select count(distinct(jobtitle)) from salaries
        where jobtitle != 'Not provided'""",conn)
```
